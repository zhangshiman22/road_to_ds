```python
data_file = os.path.join('.', 'data', 'house_large.csv')

# ==========================================
# 1. 创建包含更多行和列的原始数据集
# ==========================================
# 我们创建一个 8 行 5 列的数据集，故意在不同列制造不同数量的 NaN
raw_data = {
    'NumRooms': [3, 2, np.nan, 4, 2, np.nan, 3, 4],
    'Alley': ['Pave', np.nan, 'Grvl', np.nan, 'Pave', np.nan, 'Grvl', 'Pave'],
    'Garage': [np.nan, np.nan, np.nan, 'Attached', np.nan, 'Detached', np.nan, np.nan],  # 这一列缺失值最多 (6个 NaN)
    'YearBuilt': [2000, 2005, 2010, np.nan, 1998, 2015, 2020, 2008],
    'Price': [127500, 106000, 178100, 140000, 115000, 210000, 190000, 250000]
}

df = pd.DataFrame(raw_data)
df.to_csv(data_file, index=False)
print("--- 1. 原始数据集 ---")
print(df)
```

```python
# ==========================================
# 2. 删除缺失值最多的列
# ==========================================
# 我们可以通过 df.isna().sum() 来看每一列有多少个缺失值
# 找出缺失值最多的列名
# most_na_col = df.isna().sum().idxmax()
# print(f"各列缺失值统计：\n{df.isna().sum()}\n")
# print(most_na_col)

count = 0
count_max = 0
labels = df.columns
print(labels)
for label in labels:
    count = df[label].isna().sum()
    if count > count_max:
        count_max = count
        most_na_col = label

# 从 DataFrame 中删除该列 (axis=1 代表删除列)
df_dropped = df.drop(columns=most_na_col)

print("\n--- 2. 删除缺失列后的数据集 ---")
print(df_dropped)
```

```python
# ==========================================
# 3. 将预处理后的数据集转换为张量格式
# ==========================================
# 在转 Tensor 之前，我们需要处理剩下列的缺失值，并进行独热编码（数字化）

# 分离特征 (inputs) 和标签 (outputs，假设最后一列 Price 是标签)
inputs, outputs = df_dropped.iloc[:, :-1], df_dropped.iloc[:, -1]

# 3.1 填补数值列的缺失值（用平均值）
inputs = inputs.fillna(inputs.mean(numeric_only=True))

# 3.2 将文本列转换为 0 和 1 的 Dummy 变量（独热编码）
inputs = pd.get_dummies(inputs, dummy_na=True, dtype=float)

print("--- 3.1 预处理（数值填补+独热编码）后的 inputs ---")
print(inputs)
print("\n")

# 3.3 转换为 PyTorch 张量 (使用 to_numpy 桥梁，并指定标准 float32 类型)
X = torch.tensor(inputs.to_numpy(dtype=np.float32))
y = torch.tensor(outputs.to_numpy(dtype=np.float32))

print("--- 3.2 最终转换后的 Tensor 格式 ---")
print("X (Features) Tensor:\n", X)
print("\ny (Label) Tensor:\n", y)
```

```python
# 任务1：手写线性回归，3步走
# Step 1: 生成数据
X = torch.randn(100, 2)
true_w = torch.tensor([2.0, -3.4])
true_b = 4.2
y = X @ true_w + true_b + torch.randn(100) * 0.01 # noise
```

```python
# Step 2: 手动实现（不用nn）
w = torch.randn(2, requires_grad=True)
b = torch.zeros(1, requires_grad=True)
for epoch in range(100):
    y_hat = X @ w + b
    loss = ((y_hat - y)**2).mean()
    loss.backward()
    with torch.no_grad():
        w -= 0.03 * w.grad
        b -= 0.03 * b.grad
        w.grad.zero_()
        b.grad.zero_()
w,b
# Step 3: 用nn.Linear + SGD重写，对比两版本结果是否一致
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# ==========================================
# 3.1 定义模型、损失函数和优化器
# ==========================================
# nn.Linear(输入特征数, 输出特征数)
# 这里的输入特征数是 2，输出特征数是 1 (预测一个标量 y)
model = nn.Linear(in_features=2, out_features=1)

# 使用内置的均方误差损失 (MSELoss)
loss_fn = nn.MSELoss()

# 使用随机梯度下降优化器 (SGD)，绑定模型参数，学习率设为 0.03
optimizer = optim.SGD(model.parameters(), lr=0.03)

# ==========================================
# 3.2 训练循环 (100 个 Epoch)
# ==========================================
for epoch in range(100):
    # 1. 前向传播：计算预测值
    # 注意：nn.Linear 期待的输入 X 和 y 最好都是二维的
    # 如果你的 y 是一维的 (N,)，建议用 y.unsqueeze(1) 变成 (N, 1)
    y_pred = model(X)
    
    # 2. 计算损失
    loss = loss_fn(y_pred, y.unsqueeze(1) if y.ndim == 1 else y)
    
    # 3. 梯度清零 (相当于你手动写的 w.grad.zero_())
    optimizer.zero_grad()
    
    # 4. 反向传播：计算梯度
    loss.backward()
    
    # 5. 更新参数 (相当于你在 torch.no_grad() 下做的参数减法)
    optimizer.step()

===================
# 3.3 打印对比结果
# ==========================================
print("--- nn.Linear 版本训练结果 ---")
# model.weight 的形状默认是 (out_features, in_features)，所以需要 flatten 变成一维对比
print("训练后的 w & b:", model.weight.data.flatten())
```

```python
def use_svg_display(): #@save
    """使用svg格式在Jupyter中显示绘图"""   
    try:
        from matplotlib_inline import backend_inline
        backend_inline.set_matplotlib_formats('svg')
    except ImportError:
        # 兼容老版本或异常情况
        from IPython import display
        if hasattr(display, 'set_matplotlib_formats'):
            display.set_matplotlib_formats('svg')
            
def set_figsize(figsize=(3.5, 2.5)): #@save
    """设置matplotlib的图表大小"""
    use_svg_display()
    plt.rcParams['figure.figsize'] = figsize
```

```python
def set_axes(axes, xlabel, ylabel, xlim, ylim, xscale, yscale, legend):
    """设置matplotlib的轴"""
    axes.set_xlabel(xlabel)
    axes.set_ylabel(ylabel)
    axes.set_xscale(xscale)
    axes.set_yscale(yscale)
    axes.set_xlim(xlim)
    axes.set_ylim(ylim)
    if legend:
        axes.legend(legend)
    axes.grid()
        
```

```python
def plot(X, Y=None, xlabel=None, ylabel=None, legend=None, xlim=None, ylim=None, xscale='linear', yscale='linear',
         fmts=('-', 'm--', '--', 'g-'), figsize=(3.5, 2.5), axes=None):
    """绘制数据点"""
    if legend is None:
        legend = []
    set_figsize(figsize)
    axes = axes if axes else plt.gca()
    # 如果X有一个轴，输出True
    def has_one_axis(X):
        return (hasattr(X, "ndim") and X.ndim ==1 or isinstance(X, list) and not hasattr(X[0], "__len__"))

    if has_one_axis(X):
        X = [X]
    if Y is None:
        X, Y = [[]] * len(X), X
    elif has_one_axis(Y):
        Y = [Y]
    if len(X) != len(Y):
        X = X * len(Y)
    axes.cla()
    for x, y, fmt in zip(X, Y, fmts):
        if len(x):
            axes.plot(x, y, fmt)
        else:
            axes.plot(x,fmt)
    set_axes(axes, xlabel, ylabel, xlim, ylim, xscale, yscale, legend)
        
```

```python
# 1. 准备一堆用于画图的 x 点，必须开启 requires_grad=True
x_range = torch.arange(-5.0, 5.0, 0.1, requires_grad=True)

# 2. 计算原函数 y = sin(x)
y = torch.sin(x_range)

# 3. 核心：由于 x_range 是一维向量，PyTorch 不能直接对向量求导。
# 我们需要对 y 进行 sum() 变成标量，从而利用标准 backward 计算出每个位置各自的导数
y.sum().backward()

# 4. 拿到自动求导算出来的梯度结果 (这就是不使用 cos(x) 的秘密！)
dy_dx = x_range.grad

# 5. 转换成 numpy 数据拿去画图 (记得先用 .detach() 解绑梯度)
x_np = x_range.detach().numpy()
y_np = y.detach().numpy()
grad_np = dy_dx.numpy()

# 6. 调用你的完美高清绘图函数
plot(x_np, [y_np, grad_np], 'x', 'f(x)', 
     legend=['f(x) = sin(x)', "df/dx (Autograd)"])
```
