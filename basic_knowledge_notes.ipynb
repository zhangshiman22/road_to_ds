```python
import pandas as pd
import numpy as np
```

iloc 按位置选择，loc 按**行列名称（标签）**选择。

假设：

```python
df = pd.DataFrame(
    {"label": [5, 0, 4], "pixel1": [0, 10, 20]},
    index=["a", "b", "c"]
)
df
```

```python
# iloc：按数字位置
df.iloc[2,1]
#表示第 1 行、第 2 列，得到 0
```

```python
df.iloc[:, 1:]
#表示所有行，从第 2 列到最后。
```

```python
#注意：切片的结束位置不包含：
df.iloc[0:2]  # 只包含第0、1行
```

loc：按名称

```python
df.loc["a", "pixel1"]

# 表示索引名称为 "a" 的行、列名为 "pixel1" 的列。
```

```python
df.loc[:, "pixel1"]

#表示所有行的 "pixel1" 列。
```

```python
#注意：标签切片通常包含结束位置：

df.loc["a":"c"]  # 包含 a、b、c
```

np.random.randn() 和 np.random.rand() 不一样

np.random.rand()

生成 0～1 之间均匀分布的随机数：

- 范围：[0, 1)
- 每个范围内的数出现概率基本相同
- 平均值约为 0.5

np.random.randn()

生成服从标准正态分布的随机数：

特点：

- 平均值约为 0
- 标准差约为 1
- 本身就有正数和负数
- 数值大多在 -2～2 附近，但没有固定范围

```python
# 为什么不能用以下：
def ReLU(Z):
    if Z > 0:
        return Z
    else:
        return 0
# 正确写法：
def ReLU(Z):
    return np.maximum(Z, 0)
# 对数组里的每个元素都进行 ReLU 处理，返回一个数组。
```

Z > 0 会得到一个布尔数组：

[[True, False, ...],
 [False, True, ...]]

而 if 只能接收一个明确的 True 或 False。Python 不知道你是想判断：

数组里所有元素都大于 0？
还是只要有一个元素大于 0？

因此出现：

The truth value of an array with more than one element is ambiguous

ReLU 需要对数组中的每一个元素分别处理，可以改成：

```python
#让 softmax 更稳定，把：

def softmax(Z):
    return np.exp(Z) / sum(np.exp(Z))

# 改成：

def softmax(Z):
    Z = Z - np.max(Z, axis=0, keepdims=True)
    exp_Z = np.exp(Z)
    return exp_Z / np.sum(exp_Z, axis=0, keepdims=True)
```

对于多个样本：

Z = np.array([
    [1000, 20],
    [1002, 25],
    [1001, 21]
])

形状是：

3个类别 × 2个样本

执行：

np.max(Z, axis=0)

得到每个样本各自的最大类别分数：

[1002, 25]

为什么减去最大值不会改变 softmax？

Softmax 为：

$$ \operatorname{softmax}(z_i) = \frac{e^{z_i}}{\sum_j e^{z_j}} $$

所有分数同时减去最大值 \(c\)：

$$ \frac{e^{z_i-c}}{\sum_j e^{z_j-c}} = \frac{e^{z_i}e^{-c}}{e^{-c}\sum_j e^{z_j}} = \frac{e^{z_i}}{\sum_j e^{z_j}} $$

分子和分母中的 \(e^{-c}\) 会抵消，所以最终概率完全不变。
