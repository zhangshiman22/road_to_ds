核心思路 (Core Logic)
初始化：维护一个计数器 count = 1，并从字符串的第二个字符开始遍历。
比较相邻字符：
如果当前字符与前一个字符相同，计数器 count 加 1。
如果当前字符与前一个字符不同，说明上一个字符的连续片段结束了。我们将“前一个字符 + count”拼接到结果中，然后重置 count = 1。
处理末尾：循环结束后，别忘了把最后一组字符和它的 count 拼接到结果里。
Count = 1


```python
#空间复杂度为n
def compress_string(s: str) -> str:
    if not s:
        return "" #如果字符串为空，返回空字符串
    count = 1
    result = []
    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            count += 1
        else:
            result += s[i-1] + str(count)
            count = 1
    result.append (s[-1] + str(count))
    return  "".join(result)
       
                
```

```python
s = 'aabbcccaa'
compress_string(s)
```

```python
from typing import List

def compress_inplace(chars: List[str]) -> int:
    if not chars:
        return 0
    
    write = 0
    read = 0
    
    while read < len(chars):
        char = chars[read]
        count = 0
        
        # Count the occurrences of the current character
        while read < len(chars) and chars[read] == char:
            read += 1
            count += 1
        
        # Write the character
        chars[write] = char
        write += 1
        
        # Write the count (even if it is 1, based on your "a2b2" requirement)
        for digit in str(count):
            chars[write] = digit
            write += 1
            
    # The list is modified in-place up to index `write`
    return write

# --- Test ---
data = list("aabbcccaa")  # ['a', 'a', 'b', 'b', 'c', 'c', 'c', 'a', 'a']
new_length = compress_inplace(data)

# Print the modified part of the list
print("".join(data[:new_length]))  # Output: a2b2c3a2
```
